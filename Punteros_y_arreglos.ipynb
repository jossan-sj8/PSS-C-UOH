Ejercicio 1: implementar una función que intercambie 2 valores.


```python

%%writefile swap.c

#include <stdio.h>

void swap(int *a , int *b){
  int aux = *a;
  *a = *b;
  *b = aux;
}

int main(){
  int a=1, b=2;
  printf("Valores iniciales: a es igual a %d y b es igual a %d\n", a, b);

  swap(&a,&b);

  printf("Valores finales: a ahora es %d y b ahora es %d\n", a, b);

  return 0;
}
```

```python
!gcc swap.c -o swap && ./swap
```

Ejercicio 2: Implementar una función para crear una variable y que nos devuelva la dirección de memoria.

```python
%%writefile getvar1.c

#include <stdio.h>

int *getvar(){
  int x=1;
  return &x;
}

int main(){
  int *p = *getvar();
  printf("%d\n", p);

}
```

```python
!gcc getvar1.c -o getvar1 && ./getvar1
```

Error de segmentación por la memoria.

Solución: usando variables dinámicas

```python
%%writefile getvar.c

#include <stdio.h>
#include <stdlib.h>

int *getvar(){
  int *q = malloc(sizeof(int));
  *q = 1;
  return q;
}

int main(){
  int *p = getvar();
  printf("%d\n", *p);
  free(p);

}
```

```python
!gcc getvar.c -o getvar && ./getvar
```

La opción de (void ) para imprimir en Google colab las variables p, p, &p

```python
%%writefile getvar.c

#include <stdio.h>
#include <stdlib.h>

int *getvar(){
  int *q = malloc(sizeof(int));
  *q = 1;
  return q;
}

int main(){
  int *p = getvar();
  printf("%d\n", *((int *)p));//opcion de (void *)
  free(p);

}
```

Ejercicio usando aritmética de punteros, arreglos y strings, implemente una función que calcule el largo dado un string.


```python
%%writefile largo.c

#include <stdio.h>

int str_len(char *s) {
    int len = 0;
    while (*s != 0) {
        len++;
        s++;
    }
    return len;
}

int main() {
    printf("Largo de 'perro': %d\n", str_len("perro"));
    printf("Largo de 'programacion': %d\n", str_len("programacion"));

    return 0;
}
```
