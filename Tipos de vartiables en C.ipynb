Tipos de variables en C

Ejemplo variables tipo Globales, si no se declara el valor por defecto es 0:

```python
%%writefile global.c

#include <stdio.h>

int a;
int b = 200;

int main(){
  printf("Valor inicial de a: %d\n",a);
  printf("Valor inicial de b: %d\n",b);

  a+=1;
  printf("Valor final de a:%d\n", a);

  return 0;
}
```

```python

!gcc global.c -o global && ./global
```

Ejemplo variables de tipo local

```python

%%writefile locales.c

#include <stdio.h>

double factorial(int n){
  double p= 1.0;
  while (n>1){
    p*= n;
    n--;
  }
  return p;
}

int main(){
    printf("Factorial de 5:%lf\n", factorial(5));
    return 0;

}
```

```python
!gcc locales.c -o locales && ./locales
```
