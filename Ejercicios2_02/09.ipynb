Manipulacion de bits (bitwise Operation)

```python
%%writefile op_bit.c

#include <stdio.h>

unsigned char encender_bit(unsigned char num, int pos){
  return num | (1 << pos);
}

int consultar_bit(unsigned char num, int pos){
  return (num>>pos) & 1;
}

int main(){
  unsigned char valor = 0;

  valor = encender_bit(valor,3);
  printf("Nuevo valor:%d\n", valor);
  printf("Bit en posicion 3: %d\n",consultar_bit(valor,3));
  return 0;
}
```

memoria dinamica con malloc - free - array

```python
%%writefile mem_d.c
#include <stdio.h>
#include <stdlib.h>

int *arrglo_dinamico(int tamano){
  int *arr = (int *)malloc(tamano * sizeof(int));

  if(arr==NULL){
    printf("no teni memoria");
    return NULL;
  }

  for(int i = 0; i < tamano; i++){
    arr[i] = (i + 1) * 10;
  }

  return arr;
}


int main(){
  int n = 4;
  int *mi_arreglo = arrglo_dinamico(n);

  if(mi_arreglo != NULL){
    for(int i = 0; i < n; i++){
      printf("%d", mi_arreglo[i]);
      printf("\n");
    }

    free(mi_arreglo);
  }
  return 0;

}
```

```python
!gcc mem_d.c -o mem && ./mem
```

Implementa una función char *strcat_dynamic(const char *s1, const char *s2)

```python
%%writefile ejem_char.c
#include <stdio.h>
#include <stdlib.h>

char *strcat_dynamic(const char *s1, const char *s2) {
    int len1 = 0, len2 = 0;

    // 1. Contar longitud de s1
    while (s1[len1] != '\0') {
        len1++;
    }

    // 2. Contar longitud de s2
    while (s2[len2] != '\0') {
        len2++;
    }

    // 3. Asignación dinámica en el Heap para ambas cadenas + '\0'[cite: 5, 7]
    char *resultado = (char *)malloc((len1 + len2 + 1) * sizeof(char));

    if (resultado == NULL) {
        return NULL; // Control de error[cite: 6]
    }

    // 4. Copiar contenido de s1
    int i = 0;
    for (int j = 0; j < len1; j++, i++) {
        resultado[i] = s1[j];
    }

    // 5. Copiar contenido de s2 a continuación
    for (int j = 0; j < len2; j++, i++) {
        resultado[i] = s2[j];
    }

    // 6. Colocar el terminador nulo al final
    resultado[i] = '\0';

    return resultado; // Retornar dirección de memoria asignada[cite: 5]
}

int main() {
    const char *cadena1 = "Sistemas ";
    const char *cadena2 = "Operativos";

    // Llamada a la función que crea memoria dinámica
    char *texto_unido = strcat_dynamic(cadena1, cadena2);

    if (texto_unido != NULL) {
        printf("Resultado final: %s\n", texto_unido); // Imprime: Sistemas Operativos

        // La memoria reservada dentro de la función debe ser liberada aquí[cite: 5, 7]
        free(texto_unido);
    }

    return 0;
}
```
