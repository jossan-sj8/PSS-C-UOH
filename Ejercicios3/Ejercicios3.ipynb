#Intercambio de dos posiciones en un Arreglo

```python
%%writefile inter_array.c
#include <stdio.h>

void swap(int *a, int *b) {
    int temp = *a; // Guarda el valor en la dirección 'a'
    *a = *b;       // Sobrescribe el valor en 'a' con el de 'b'
    *b = temp;     // Escribe 'temp' en la dirección 'b'
}

int main() {
    int numeros[5] = {10, 20, 30, 40, 50};
    int i = 1, j = 3; // Queremos intercambiar el 20 (índice 1) y el 40 (índice 3)

    printf("Antes del swap:  numeros[1] = %d, numeros[3] = %d\n", numeros[i], numeros[j]);

    // Enfriamos la dirección de memoria de cada posición usando '&'
    swap(&numeros[i], &numeros[j]);

    printf("Después del swap: numeros[1] = %d, numeros[3] = %d\n", numeros[i], numeros[j]);

    return 0;
}
```

```python
!gcc inter_array.c -o inter_array && ./inter_array
```

# Intercambio de valores entre Nodos de una Lista Enlazada (swap_nodos)

```python
%%writefile inter_node.c
#include <stdio.h>
#include <stdlib.h>

// Definición del nodo para la lista enlazada
typedef struct Nodo {
    int valor;
    struct Nodo *siguiente;
} Nodo;

void swap(int *a, int *b) {
    int temp = *a;
    *a = *b;
    *b = temp;
}

int main() {
    // Creamos dos nodos dinámicamente en el Heap
    Nodo *nodo1 = (Nodo *)malloc(sizeof(Nodo));
    Nodo *nodo2 = (Nodo *)malloc(sizeof(Nodo));

    nodo1->valor = 100;
    nodo1->siguiente = nodo2;

    nodo2->valor = 500;
    nodo2->siguiente = NULL;

    printf("Antes: Nodo 1 = %d, Nodo 2 = %d\n", nodo1->valor, nodo2->valor);

    // Le pasamos las direcciones de la propiedad 'valor' de cada nodo
    swap(&(nodo1->valor), &(nodo2->valor));

    printf("Después: Nodo 1 = %d, Nodo 2 = %d\n", nodo1->valor, nodo2->valor);

    // Liberación de memoria
    free(nodo1);
    free(nodo2);

    return 0;
}
```

```python
!gcc inter_node.c -o inter_node && ./inter_node
```

# Inversión de una Cadena In-Place

```python
%%writefile inver_in_place.c
#include <stdio.h>

void reverse_string(char *str) {
    if (!str) return;

    // 1. Calcular longitud manualmente
    int len = 0;
    while (str[len] != '\0') {
        len++;
    }

    // 2. Intercambiar extremos avanzando hacia el centro
    int i = 0, j = len - 1;
    while (i < j) {
        char temp = str[i];
        str[i] = str[j];
        str[j] = temp;
        i++;
        j--;
    }
}

int main() {
    char texto[] = "Minato";
    reverse_string(texto);
    printf("Texto invertido: %s\n", texto); // Salida
    return 0;
}
```

# Manejo de cadenas estáticas(my_strcat)

```python
%%writefile cad_sta.c
#include <stdio.h>

char *my_strcat(char *dest, const char *src) {
    char *ptr = dest;

    // 1. Desplazar el puntero al final de 'dest'
    while (*ptr != '\0') {
        ptr++;
    }

    // 2. Copiar los caracteres de 'src' a continuación
    while (*src != '\0') {
        *ptr = *src;
        ptr++;
        src++;
    }

    // 3. Añadir terminador nulo final
    *ptr = '\0';
    return dest;
}

int main() {
    char buffer[50] = "Hola ";
    my_strcat(buffer, "Mundo");
    printf("Resultado: %s\n", buffer); // Salida
    return 0;
}
```

# Concatenación Dinámica en el Heap

```python
%%writefile conc_dina.c
#include <stdio.h>
#include <stdlib.h>

char *strcat_dynamic(const char *s1, const char *s2) {
    int len1 = 0, len2 = 0;

    while (s1[len1] != '\0') len1++;
    while (s2[len2] != '\0') len2++;

    // Reservar memoria para len1 + len2 + 1 ('\0')
    char *res = (char *)malloc((len1 + len2 + 1) * sizeof(char));
    if (!res) return NULL;

    int i = 0;
    for (int j = 0; j < len1; j++, i++) res[i] = s1[j];
    for (int j = 0; j < len2; j++, i++) res[i] = s2[j];
    res[i] = '\0';

    return res;
}

int main() {
    char *str = strcat_dynamic("SO - ", "UOH");
    if (str != NULL) {
        printf("%s\n", str);
        free(str); // Liberación obligatoria
    }
    return 0;
}
```

# Conteo Recursivo de Nodos Hoja en un Árbol Binario

```python
%%writefile con_hojas.c
#include <stdio.h>
#include <stdlib.h>

typedef struct Node {
    int val;
    struct Node *left;
    struct Node *right;
} Node;

Node* create_node(int val) {
    Node *n = (Node *)malloc(sizeof(Node));
    n->val = val;
    n->left = NULL;
    n->right = NULL;
    return n;
}

int contar_hojas(Node *root) {
    if (root == NULL) return 0;

    // Un nodo hoja no tiene hijos a la izquierda ni a la derecha
    if (root->left == NULL && root->right == NULL) return 1;

    return contar_hojas(root->left) + contar_hojas(root->right);
}

int main() {
    Node *root = create_node(10);
    root->left = create_node(5);
    root->right = create_node(15);
    root->left->left = create_node(2); // Hoja 1
    root->left->right = create_node(7); // Hoja 2

    printf("Hojas totales: %d\n", contar_hojas(root));
    return 0;
}
```

# Evaluador de Expresiones Matemáticas con Árbol Binario

```python
%%writefile exp_arbol.c
#include <stdio.h>
#include <stdlib.h>

typedef struct Node {
    char op;       // Operador '+', '*' o '\0' si es operando
    int val;       // Valor numérico si es operando
    struct Node *left;
    struct Node *right;
} Node;

int evaluar_arbol(Node *root) {
    if (!root) return 0;

    // Si es un nodo hoja, retorna directamente su valor entero
    if (!root->left && !root->right) return root->val;

    int izq = evaluar_arbol(root->left);
    int der = evaluar_arbol(root->right);

    if (root->op == '+') return izq + der;
    if (root->op == '*') return izq * der;

    return 0;
}

int main() {
    // Expresión representada: (3 + 5) * 2
    Node *root = (Node *)malloc(sizeof(Node)); root->op = '*';
    root->left = (Node *)malloc(sizeof(Node)); root->left->op = '+';
    root->right = (Node *)malloc(sizeof(Node)); root->right->op = '\0'; root->right->val = 2;
    root->right->left = root->right->right = NULL;

    root->left->left = (Node *)malloc(sizeof(Node)); root->left->left->op = '\0'; root->left->left->val = 3;
    root->left->left->left = root->left->left->right = NULL;

    root->left->right = (Node *)malloc(sizeof(Node)); root->left->right->op = '\0'; root->left->right->val = 5;
    root->left->right->left = root->left->right->right = NULL;

    printf("Resultado de la expresión: %d\n", evaluar_arbol(root));
    return 0;
}
```
